# Dynamic Pricing Opportunity Scanner

This notebook loads the synthetic pricing dataset, checks data quality, calculates KPIs, estimates price elasticity using a log-log regression, and reviews product-level pricing opportunities.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

df = pd.read_csv("../data/pricing_sales_data.csv")
print(df.shape)
print(df.isna().sum())

model_df = df[(df.listed_price > 0) & (df.units_sold > 0)].copy()
X = np.log(model_df[["listed_price"]])
y = np.log(model_df["units_sold"])
model = LinearRegression().fit(X, y)
print("Estimated price elasticity:", model.coef_[0])
print("R-squared:", model.score(X, y))

opps = pd.read_csv("../docs/product_pricing_opportunities.csv")
opps.head(15)
```
